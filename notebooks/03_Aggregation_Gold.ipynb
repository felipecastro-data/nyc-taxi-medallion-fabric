# 🥇 NYC Taxi Analytics - Gold Layer
> **Goal:** Transform cleaned silver data into a high-performance **Summary Table** for business insights. 📈

### 🛠️ Engineering Steps:
1. **Load** the cleaned data from the Silver layer.
2. **Aggregate** metrics: Total trips, average fare, and total tips per day and per vendor. 🚕💰
3. **Overwrite/Merge** into the Gold table for reporting.

```python
from pyspark.sql.functions import col, date_format, sum, avg, count

# 1. Load Data from Silver
df_silver = spark.read.table("silver_nyc_taxi")

# 2. Create Business Aggregations
# Grouping by Date and Vendor to see performance trends
df_gold_summary = df_silver.groupBy(
    date_format(col("tpep_pickup_datetime"), "yyyy-MM-dd").alias("pickup_date"),
    col("VendorID")
).agg(
    count("*").alias("total_trips"),
    sum("total_amount").alias("total_revenue"),
    avg("trip_distance").alias("avg_distance"),
    sum("tip_amount").alias("total_tips")
).orderBy("pickup_date")

# 3. Write to Gold Table
# For summary tables, 'overwrite' is common to ensure we have the latest totals
target_gold_table = "gold_taxi_daily_summary"
df_gold_summary.write.format("delta").mode("overwrite").saveAsTable(target_gold_table)

print(f"✅ Gold Summary Table '{target_gold_table}' is ready for reporting!")
display(df_gold_summary.limit(10))
```
